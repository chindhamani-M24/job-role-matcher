```python
# ==========================================
# Interactive Job Role Matcher
# Google Colab Copy-Paste Ready Script
# ==========================================

import subprocess
import sys
import pandas as pd

def install_package(package):
    try:
        __import__(package)
    except ImportError:
        subprocess.check_call([sys.executable, "-m", "pip", "install", package])

print("--- Checking and installing required libraries ---")
install_package("pandas")
install_package("scikit-learn")

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

# 1. Training Dataset (Background descriptions labeled with Job Roles)
data = {
    'profile_text': [
        "Experienced in Python, SQL, machine learning algorithms, data modeling, and pandas.",
        "Built scalable web applications using JavaScript, React, Node.js, and REST APIs.",
        "Managed product backlogs, user stories, sprint planning, and cross-functional teams.",
        "Created social media campaigns, SEO strategies, Google Ads, and email marketing.",
        "Designed databases, wrote complex SQL queries, and optimized server backend code.",
        "Conducted user research, wireframed UI/UX designs in Figma, and tested prototypes."
    ],
    'job_role': [
        'Data Scientist',
        'Software Engineer',
        'Product Manager',
        'Digital Marketer',
        'Software Engineer',
        'UX/UI Designer'
    ]
}

df = pd.DataFrame(data)

print("\n--- Training Machine Learning Job Role Classifier ---")
# Build and train classification pipeline
model = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('clf', LogisticRegression())
])
model.fit(df['profile_text'], df['job_role'])

# 2. Interactive Loop for Google Colab
print("\n" + "="*60)
print("💼 INTERACTIVE JOB ROLE MATCHER READY")
print("Type a short professional background or skill set below.")
print("Type 'exit' or 'quit' to stop.")
print("="*60 + "\n")

while True:
    try:
        user_input = input("Enter your background/skills: ").strip()
    except EOFError:
        break

    if user_input.lower() in ['exit', 'quit']:
        print("\nExiting job role matcher. Have a great day!")
        break

    if not user_input:
        print("⚠️ Please enter valid text.\n")
        continue

    # Predict job role
    prediction = model.predict([user_input])[0]

    print(f"\n[AI Analysis]")
    print(f"👉 Recommended Job Role: {prediction}")
    print("-" * 60 + "\n")
```
